# Try the API Client

The notebook connects to the deployed API, authenticates with an API key and retrieves watch market data from the live backend. I wanted this notebook to make the API side of the project easy to test without needing to run the web application.

## API Access

The API is protected with an API key. If you would like to run the notebook yourself, please contact me directly and I can provide an API key.

Note: The key is entered at runtime and is not stored in the notebook.

```python
from getpass import getpass

import pandas as pd
import requests

BASE_URL = "https://api.dcanguven.workers.dev"
API_KEY = getpass("API key: ")

HEADERS = {
    "X-API-Key": API_KEY,
    "Accept": "application/json",
}
```

## Query Watch Records

This example filters for Rolex and sorts the results by median market price. The same endpoint also supports pagination and additional filters.

```python
params = {
    "brand": "Rolex",
    "page": 1,
    "page_size": 10,
    "sort": "price_median",
    "order": "desc",
}

response = requests.get(
    f"{BASE_URL}/api/v1/watches",
    headers=HEADERS,
    params=params,
    timeout=10,
)

response.raise_for_status()

payload = response.json()
payload
```

## A More Readable View

```python
watches = pd.DataFrame(payload["data"])

columns = [
    "id",
    "brand",
    "model",
    "price_median",
    "year_from",
    "year_to",
]

watches[columns]
```

## Retrieve a Full Record

```python
watch_id = payload["data"][0]["id"]

response = requests.get(
    f"{BASE_URL}/api/v1/watches/{watch_id}",
    headers=HEADERS,
    timeout=10,
)

response.raise_for_status()

watch = response.json()
watch
```
